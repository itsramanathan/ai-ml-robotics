```python
import numpy as np

A = np.array([[2, 1],
              [1, 3]])

A_inv = np.linalg.inv(A)

print("A =")
print(A)
print("\nA_inv =")
print(A_inv)
print("\nA @ A_inv = ")
print(A @ A_inv)
```

```python
S = np.array([[2, 4],
              [1, 2]])

det_S = np.linalg.det(S)
print(f"det(S) = {det_S}")

try:
    S_inv = np.linalg.inv(S)
    print(f"S_inv = {S_inv}")
except np.linalg.LinAlgError as e:
    print(f"Expected Error: {e}")
```

```python
print(f"rank(A) = {np.linalg.matrix_rank(A)}")
print(f"rank(S) = {np.linalg.matrix_rank(S)}")
```

```python
from scipy.linalg import null_space

ns_A = null_space(A)
ns_S = null_space(S)

print(f"Null space of A:\n{ns_A}")
print(f"Shape: {ns_A.shape}")

print(f"Null space of S:\n{ns_S}")
print(f"Shape: {ns_S.shape}")

print(f"\nVerify S crushes its null vector to zero:")
print(f"S @ ns_S = {(S @ ns_S).flatten()}")
```

```python
M = np.array([[1, 0, 1],
              [0, 1, 1],
              [0, 0, 0]])

print("M =")
print(M)
print(f"\nM has 3 columns but rank(M) = {np.linalg.matrix_rank(M)}")
print(f"So ColSpace(M) is {np.linalg.matrix_rank(M)}-dimensional, not 3-dimensional")

target_inside = np.array([3, 5, 0])   # in xy-plane → reachable
target_outside = np.array([3, 5, 1])  # has z-component → unreachable

# For "inside" target: solve Mx = target_inside (least-squares since rank-deficient)
x_inside, residuals_inside, _, _ = np.linalg.lstsq(M, target_inside, rcond=None)
print(f"\nReaching [3,5,0]: M @ x = {M @ x_inside}")
print(f"Match? {np.allclose(M @ x_inside, target_inside)}")

x_outside, residuals_outside, _, _ = np.linalg.lstsq(M, target_outside, rcond=None)
print(f"\nReaching [3,5,1]: M @ x = {M @ x_outside}")
print(f"Match? {np.allclose(M @ x_outside, target_outside)}")
```
